## 1.1 Asignación y actualización
`=` asocia un nombre con un objeto. Los operadores `+=`, `-=`, `*=`, `/=`, `//=`, `%=` y `**=` permiten actualizar una variable. Con números, `x += y` puede explicarse como `x = x + y`; con objetos mutables hay diferencias que estudiaremos al trabajar con listas.

| Operador | Ejemplo | Resultado si inicialmente `x = 10` |
|---|---|---|
| `+=` | `x += 3` | `13` |
| `-=` | `x -= 3` | `7` |
| `*=` | `x *= 3` | `30` |
| `/=` | `x /= 4` | `2.5` |
| `//=` | `x //= 4` | `2` |
| `%=` | `x %= 4` | `2` |
| `**=` | `x **= 2` | `100` |

Cada fila de la tabla parte de `x = 10`, no del resultado de la fila anterior.

## 1.2 Comparaciones, pertenencia e identidad
Las comparaciones habituales son `==`, `!=`, `<`, `<=`, `>` y `>=`. Con números producen booleanos (`True` o `False`). No confundas `=` con `==`.

- `in` y `not in` comprueban pertenencia; en una cadena buscan subcadenas y en un diccionario buscan claves.
- `==` compara valores según el tipo de los objetos.
- `is` compara identidad: pregunta si dos referencias apuntan al mismo objeto. No se usa para comparar calificaciones, cadenas o enteros. Una aplicación habitual es `resultado is None`.

### Igualdad no es identidad

Dos listas distintas pueden contener los mismos valores.

````python
lista_a = [8, 9]
lista_b = [8, 9]
lista_c = lista_a
print(lista_a == lista_b)
print(lista_a is lista_b)
print(lista_a is lista_c)
print("ACT002" in ["ACT001", "ACT002"])
print("riesgo" in "análisis de riesgo")
````

# 2. Listas, tuplas y rangos
## 2.1 Elegir una estructura
Una secuencia tiene orden y permite acceso por posición. «Ordenada» no significa «clasificada de menor a mayor».

| Tipo | Uso inicial | ¿Mutable? | ¿Secuencia? |
|---|---|---|---|
| `str` | Texto | No | Sí |
| `list` | Registros que se modifican | Sí | Sí |
| `tuple` | Agrupación de campos | No | Sí |
| `range` | Progresión de enteros | No | Sí |
| `set` | Elementos distintos y operaciones de conjuntos | Sí | No |
| `dict` | Asociaciones de claves con valores | Sí | No |

Los diccionarios conservan el orden de inserción, pero no son secuencias indexadas por posición. Los conjuntos no tienen un orden de posiciones utilizable.

### Índices y rebanadas de listas

El índice inicial es 0; el límite final de una rebanada se excluye.

````python
montos = [1200, 800, 3500, 500]
print(montos[0])
print(montos[-1])
print(montos[1:3])
print(montos[::-1])
montos[1] = 900
print(montos)
# montos[10] produciría IndexError.
print(montos[10:])
````

### Asignación a una rebanada

Una rebanada puede reemplazarse por una cantidad distinta de elementos.

````python
valores = [0, 1, 2, 3, 4, 5]
valores[:3] = [0, 0, 0]
print(valores)
valores[1:3] = [9]
print(valores)
````

## 2.3 `range(inicio, fin, paso)`
Representa una progresión de enteros sin almacenar todos sus elementos. El inicio predeterminado es 0, el paso predeterminado es 1 y el fin se excluye. El paso puede ser negativo, pero no cero.

Un rango puede estar vacío cuando los límites y el signo del paso no permiten avanzar. No todo argumento incorrecto produce un rango vacío: un paso cero produce `ValueError` y un argumento no entero, como `2.5`, produce `TypeError`.

# 3. Ciclos: repetir con un propósito
## 3.1 `for`
Recorre los elementos de un iterable. Termina cuando se agotan, salvo interrupción. No exige que conozcas previamente cuántos elementos habrá; en los primeros ejemplos utilizamos secuencias finitas y conocidas.

### Contador y acumulador

El contador registra cuántos casos cumplen una condición; el acumulador suma sus importes.

````python
montos = [1200, 0, 800, 2500]
conteo = 0
acumulado = 0
for monto in montos:
    if monto > 0:
        conteo += 1
        acumulado += monto
print("Registros positivos:", conteo)
print("Total:", acumulado)
if conteo > 0:
    print("Promedio de los positivos:", acumulado / conteo)
````

### Distinguir la media global y la media condicionada

Una media depende del conjunto de observaciones que decidas incluir.

````python
montos = [1200, 0, 800, 2500]
print("Media de todos los registros:", sum(montos) / len(montos))
positivos = []
for monto in montos:
    if monto > 0:
        positivos.append(monto)
print("Media entre los positivos:", sum(positivos) / len(positivos))
````

## 3.2 `while`
Repite un bloque mientras la condición sea verdadera. Antes de escribirlo identifica: estado inicial, condición, actualización y terminación. Si no cambia nada que permita terminar, puede producirse un ciclo infinito.

### Capitalización compuesta como modelo didáctico

Tasa fija por periodo; sin aportaciones, retiros, impuestos ni comisiones.

````python
capital = 1000.0
tasa = 0.05
periodo = 0
while periodo < 3:
    capital *= 1 + tasa
    periodo += 1
    print("Periodo {}: ${:.2f}".format(periodo, capital))
````

## 3.3 `break`, `continue` y `pass`
- `break` termina el ciclo más interno en el que se ejecuta.
- `continue` omite el resto de la iteración actual y comienza la siguiente.
- `pass` no altera el flujo: no hace nada.

En un `while`, colocar la actualización después de un `continue` puede impedir que el estado avance. Evita modificar el tamaño de una lista mientras la recorres sin una estrategia explícita.

### Buscar el primer monto que supera un umbral

La variable `encontrado` guarda un resultado o permanece en None.

````python
montos = [100, 450, 1700, 900]
encontrado = None
for monto in montos:
    if monto > 1000:
        encontrado = monto
        break
if encontrado is None:
    print("No se encontró un monto mayor que 1000")
else:
    print("Primer monto mayor que 1000:", encontrado)
````

### Descartar registros negativos

Un dato negativo se omite en este ejercicio. En una aplicación real podría ser un ajuste válido: la regla depende del modelo de datos.

````python
montos = [100, -1, 250, -5, 300]
total = 0
for monto in montos:
    if monto < 0:
        continue
    total += monto
print(total)
````

````python
matriz = [[10, 20], [30, 40], [50, 60]]
for fila in matriz:
    total_fila = 0
    for valor in fila:
        total_fila += valor
    print("Total de fila:", total_fila)
````

# 4. Herramientas para listas y recorridos coordinados
## 4.1 Métodos que modifican listas

| Método | Efecto | Resultado devuelto |
|---|---|---|
| `append(x)` | Agrega x como un elemento | `None` |
| `extend(iterable)` | Agrega los elementos del iterable | `None` |
| `insert(i, x)` | Inserta x antes de i | `None` |
| `remove(x)` | Elimina la primera coincidencia; error si no existe | `None` |
| `pop(i)` | Elimina en i; por defecto el último | Elemento eliminado |
| `clear()` | Vacía la lista | `None` |
| `reverse()` | Invierte la lista | `None` |
| `sort()` | Ordena la lista en su lugar | `None` |

Otros métodos: `count(x)` cuenta coincidencias, `index(x)` devuelve el primer índice o produce `ValueError`, y `copy()` devuelve una copia superficial.

**No confundir:** `lista = lista.sort()` reemplaza la variable por `None`. Para una lista nueva utiliza `sorted(lista)`.

### append no es extend

append agrega un objeto; extend recorre otro iterable.

````python
lista_a = [100, 200]
lista_a.append([300, 400])
lista_b = [100, 200]
lista_b.extend([300, 400])
print(lista_a)
print(lista_b)
extraido = lista_b.pop()
print("Extraído:", extraido)
print("Lista restante:", lista_b)
````

### Ordenar con y sin modificación

Los números se ordenan en este ejemplo con su orden habitual.

````python
montos = [300, 100, 200]
ordenados = sorted(montos)
print("Original:", montos)
print("Nueva lista:", ordenados)
resultado = montos.sort(reverse=True)
print("Orden descendente:", montos)
print("Resultado de sort:", resultado)
````

## 4.2 Funciones incorporadas y sentencia `del`
`len`, `min`, `max`, `sum`, `sorted`, `list`, `enumerate`, `all` y `any` son funciones, no métodos de lista. `del` es una sentencia.

`sum([])` devuelve 0; `min([])` y `max([])` producen error si no se proporciona una alternativa. `all([])` es `True` y `any([])` es `False`. Estos comportamientos importan al validar colecciones vacías.

````python
valores = [5, 8, 3]
print(len(valores), min(valores), max(valores), sum(valores))
print(all([True, True, False]))
print(any([False, True, False]))
print(all([]), any([]))
del valores[1]
print(valores)
````

## 4.3 Alias y copias
`b = a` no copia una lista: crea otra referencia al mismo objeto. `a.copy()` copia la lista externa, pero los objetos anidados continúan compartidos. Para estructuras profundas se necesita otra estrategia, por ejemplo `copy.deepcopy()`; se presenta como ampliación.

Con listas, `+=` suele modificar el objeto existente; `a = a + [...]` crea una lista nueva. Por eso la equivalencia de asignación aumentada explicada para números no debe generalizarse sin matices.

### Copiar y compartir referencias

Observa el efecto sobre una lista simple y después sobre una lista anidada.

````python
original = [1, 2]
alias = original
copia = original.copy()
alias.append(3)
print(original, copia)
matriz = [[1, 2], [3, 4]]
superficial = matriz.copy()
superficial[0][0] = 99
print(matriz)
````

# 5. Conjuntos y diccionarios
## 5.1 Conjuntos (`set`)
Un conjunto almacena elementos distintos y no ofrece posiciones indexadas. Es mutable, pero sus elementos deben ser **hashables**: por ejemplo, cadenas, enteros y tuplas cuyos elementos también lo sean. Las listas y los diccionarios no son hashables. No basta decir que cualquier objeto inmutable sirve.

`{}` crea un diccionario vacío; para un conjunto vacío utiliza `set()`.

| Operación | Símbolo | Método |
|---|---|---|
| Unión | `A | B` | `A.union(B)` |
| Intersección | `A & B` | `A.intersection(B)` |
| Diferencia | `A - B` | `A.difference(B)` |
| Diferencia simétrica | `A ^ B` | `A.symmetric_difference(B)` |
| Subconjunto | `A <= B` | `A.issubset(B)` |
| Superconjunto | `A >= B` | `A.issuperset(B)` |

No presupongas un orden al convertir un conjunto a lista o tupla. Utiliza `sorted()` si necesitas mostrar cadenas o números comparables con un orden estable.

### Estudiantes que practican deportes

Aplicación conservada de las notas originales, con datos ficticios simplificados.

````python
futbol = {"Ana", "Luis", "María", "José"}
tenis = {"Ana", "José", "Mateo"}
atletismo = {"Ana", "Luis", "Sofía"}
print("Los tres:", sorted(futbol & tenis & atletismo))
print("Fútbol y tenis, no atletismo:", sorted((futbol & tenis) - atletismo))
print("Al menos un deporte:", sorted(futbol | tenis | atletismo))
solo_uno = (futbol - tenis - atletismo) | (tenis - futbol - atletismo) | (atletismo - futbol - tenis)
print("Exactamente uno:", sorted(solo_uno))
````

### Métodos que modifican conjuntos
`add(x)` agrega un elemento; `update(iterable)` incorpora varios. `remove(x)` produce error si x no existe; `discard(x)` no. `pop()` elimina y devuelve un elemento **arbitrario**, no una muestra aleatoria; en un conjunto vacío produce `KeyError`. Los métodos `difference_update`, `intersection_update` y `symmetric_difference_update` modifican el conjunto en lugar de crear otro.

### Pertenencia y cambios

El resultado se presenta ordenado sin imponer orden al conjunto original.

````python
folios = {"SIN001", "SIN002"}
folios.add("SIN003")
folios.update(["SIN003", "SIN004"])
folios.discard("SIN999")
print(sorted(folios))
print("SIN002" in folios)
print({"SIN001"}.issubset(folios))
print(folios.isdisjoint({"SIN900"}))
````

### Palabra pentavocálica: resultado siempre booleano

Regla de este ejercicio: una sola palabra alfabética sin espacios; se comprueban a, e, i, o, u sin normalizar tildes. No es una clasificación lingüística exhaustiva.

````python
def es_pentavocalica(texto):
    palabra = texto.strip().lower()
    if not palabra.isalpha():
        return False
    for vocal in "aeiou":
        if vocal not in palabra:
            return False
    return True

print(es_pentavocalica("murcielago"))
print(es_pentavocalica("actuario"))
print(es_pentavocalica("dos palabras"))
````

### Dos modelos de tarifa ficticia

No confundir cobro proporcional con cobro por tramo iniciado. Se suponen distancias no negativas. Los importes no son tarifas reales de Huatulco.

````python
import math

def tarifa_proporcional(distancia_km, base=5.0, por_tramo=2.5):
    """Cobro proporcional a tramos de 120 metros."""
    return base + por_tramo * distancia_km * 1000 / 120

def tarifa_por_tramo(distancia_km, base=5.0, por_tramo=2.5):
    """Cobro por cada tramo de 120 metros iniciado."""
    tramos = math.ceil(distancia_km * 1000 / 120)
    return base + por_tramo * tramos

for distancia in (0, 0.06, 0.12, 0.18):
    print("{:.2f} km: proporcional ${:.2f}; por tramos ${:.2f}".format(
        distancia, tarifa_proporcional(distancia), tarifa_por_tramo(distancia)))
````

## 7.3 Estado de cuenta ficticio
Formato documentado del archivo de práctica:

```text

```

La primera línea contiene el saldo inicial. Las siguientes usan `Deposito` o `Retiro`, sin acento en las etiquetas de intercambio. Los montos deben ser no negativos y usar punto decimal; las líneas vacías se omiten. El saldo final puede ser negativo: aquí no se modela una política de autorización de retiros. Se rechazan operaciones desconocidas para no ignorar datos silenciosamente.

### Reconstruir el saldo

Se usa un saldo inicial en centavos exactos y montos pequeños; para sistemas financieros reales se requieren decisiones de representación y redondeo adicionales.

````python
def calcular_saldo_archivo(ruta):
    with open(ruta, "r", encoding="utf-8") as archivo:
        encabezado = archivo.readline().strip().split("|")
        if len(encabezado) != 2 or encabezado[0] != "Saldo inicial":
            raise ValueError("Encabezado no válido")
        saldo = float(encabezado[1])
        for numero_linea, linea in enumerate(archivo, start=2):
            if not linea.strip():
                continue
            campos = linea.strip().split("|")
            if len(campos) != 2:
                raise ValueError("Formato no válido en línea {}".format(numero_linea))
            operacion, texto_monto = campos
            operacion = operacion.strip()
            monto = float(texto_monto)
            if monto < 0:
                raise ValueError("Monto negativo en línea {}".format(numero_linea))
            if operacion == "Deposito":
                saldo += monto
            elif operacion == "Retiro":
                saldo -= monto
            else:
                raise ValueError("Operación desconocida: " + operacion)
    return saldo

saldo_final = calcular_saldo_archivo("estado_cuenta.txt")
print("Saldo final: ${:.2f}".format(saldo_final))
assert math.isclose(saldo_final, 2124.5)
````

## 8.1 Modelo y reproducibilidad
El módulo `random` genera números pseudoaleatorios. Una semilla permite repetir una secuencia en un entorno compatible; no elimina la variación muestral del modelo ni garantiza secuencias idénticas entre todas las versiones de Python. No se usa aquí para contraseñas ni seguridad.

### Frecuencia relativa en una moneda sesgada

Modelo ficticio: P(Águila)=0.6, P(Sol)=0.4. Cada repetición usa la misma distribución.

````python
generador = random.Random(2026)
n_simulaciones = 5000
conteo_aguilas = 0
for _ in range(n_simulaciones):
    resultado = generador.choices(["Sol", "Águila"], weights=[0.4, 0.6], k=1)[0]
    if resultado == "Águila":
        conteo_aguilas += 1
frecuencia = conteo_aguilas / n_simulaciones
print("Águilas:", conteo_aguilas)
print("Frecuencia relativa: {:.4f}".format(frecuencia))
print("Probabilidad del modelo: 0.6000")
````

### Tiempo hasta tres resultados iguales consecutivos

Se conserva el problema del original. Moneda equilibrada e intentos independientes. Se esperan tres resultados iguales, sean Águila o Sol; no específicamente tres Águilas.

````python
def esperar_tres_iguales(generador):
    anterior = None
    consecutivos = 0
    lanzamientos = 0
    while consecutivos < 3:
        actual = generador.choice(["Águila", "Sol"])
        lanzamientos += 1
        if actual == anterior:
            consecutivos += 1
        else:
            anterior = actual
            consecutivos = 1
    return lanzamientos

generador = random.Random(2026)
repeticiones = 3000
total_lanzamientos = 0
for _ in range(repeticiones):
    total_lanzamientos += esperar_tres_iguales(generador)
print("Promedio simulado: {:.3f}".format(total_lanzamientos / repeticiones))
````

## 8.2 Monty Hall
Reglas del modelo: tres puertas; un premio colocado uniformemente; elección inicial uniforme; el presentador conoce el premio y abre siempre una puerta no elegida sin premio; si tiene dos opciones, escoge uniformemente; siempre ofrece cambiar. Se comparan las estrategias «conservar» y «cambiar».

La puerta abierta no es una elección independiente entre las restantes: depende del premio y de la elección inicial. Bajo estas reglas, conservar gana cuando la primera elección acierta (probabilidad 1/3) y cambiar gana cuando falla (probabilidad 2/3).

### Comparar estrategias con las mismas partidas

Aquí las listas candidatas se construyen mediante for para no exigir comprensiones antes de tiempo.

````python
def simular_monty_hall(n, semilla=2026):
    if n <= 0:
        raise ValueError("n debe ser positivo")
    generador = random.Random(semilla)
    gana_conservando = 0
    gana_cambiando = 0
    puertas = [1, 2, 3]
    for _ in range(n):
        premio = generador.choice(puertas)
        inicial = generador.choice(puertas)
        candidatas = []
        for puerta in puertas:
            if puerta != inicial and puerta != premio:
                candidatas.append(puerta)
        abierta = generador.choice(candidatas)
        nueva = None
        for puerta in puertas:
            if puerta != inicial and puerta != abierta:
                nueva = puerta
        if inicial == premio:
            gana_conservando += 1
        if nueva == premio:
            gana_cambiando += 1
    assert gana_conservando + gana_cambiando == n
    return gana_conservando / n, gana_cambiando / n

conservar, cambiar = simular_monty_hall(5000)
print("Conservar: {:.4f}".format(conservar))
print("Cambiar: {:.4f}".format(cambiar))
````

### Ciclo y comprensión equivalentes

````python
montos = [100, -1, 300, 100]
validos_ciclo = []
for monto in montos:
    if monto >= 0:
        validos_ciclo.append(monto)
validos = [monto for monto in montos if monto >= 0]
cuadrados = [numero**2 for numero in range(5)]
importes_distintos = {monto for monto in montos if monto >= 0}
print(validos_ciclo, validos)
print(cuadrados)
print(sorted(importes_distintos))
assert validos_ciclo == validos
````

## 9.2 Cantidad variable de argumentos
`*args` reúne argumentos posicionales en una tupla; `**kwargs` reúne argumentos nombrados en un diccionario. Los nombres son convenciones, pero los asteriscos determinan el comportamiento. No se necesitan para funciones con un número pequeño y conocido de parámetros.

### Sumar montos y mostrar campos

La función de suma devuelve 0 cuando no recibe argumentos.

````python
def sumar_montos(*montos):
    acumulado = 0
    for monto in montos:
        acumulado += monto
    return acumulado

def mostrar_campos(**campos):
    for clave, valor in campos.items():
        print("{}: {}".format(clave, valor))

print(sumar_montos(100, 200, 350))
print(sumar_montos())
mostrar_campos(matricula="ACT0042", promedio=8.7)
````

## 9.3 `lambda`, `map()` y `filter()`
`lambda` define una función mediante una expresión. `map(funcion, iterable)` transforma; `filter(predicado, iterable)` conserva elementos cuyo predicado tiene valor verdadero. Devuelven iteradores: aquí se convierten a lista para observar los resultados.

Evita presentar `[lambda x: x**2 for x in range(5)]` como una lista de cuadrados: crea cinco funciones iguales. El parámetro de la lambda no conserva el número del recorrido. Las capturas de variables en funciones creadas dentro de ciclos son un tema posterior.

### Tres maneras de transformar datos

Compara legibilidad antes de elegir una herramienta.

````python
def cuadrado(numero):
    return numero**2

numeros = [1, 2, 3, 4]
print(list(map(cuadrado, numeros)))
print(list(map(lambda numero: numero**2, numeros)))
print([numero**2 for numero in numeros])
print(list(filter(lambda numero: numero % 2 == 0, numeros)))
````

## 9.4 `ord()`, `chr()` y una regla ficticia de composición
`ord()` devuelve el punto de código Unicode de un carácter y `chr()` realiza la conversión inversa. Unicode no se limita a los códigos de letras inglesas. Los intervalos ASCII A–Z y a–z no incluyen letras como Ñ o á.

Una regla como «contiene una mayúscula, una minúscula y un dígito» sirve para practicar recorridos, pero no mide por sí sola la seguridad de una contraseña. No pidas contraseñas reales al alumnado. Para generar secretos reales no debe utilizarse `random`.

### Inspección de caracteres y regla didáctica

isdigit() reconoce más caracteres que los convertibles mediante int(); aquí se comprueba composición de texto, no números.

````python
print(ord("A"), chr(65))
print(ord("Ñ"), chr(ord("Ñ")))

def cumple_regla_didactica(texto):
    mayuscula = False
    minuscula = False
    digito = False
    for caracter in texto:
        if caracter.isupper():
            mayuscula = True
        if caracter.islower():
            minuscula = True
        if caracter.isdigit():
            digito = True
    return mayuscula and minuscula and digito

print(cumple_regla_didactica("Ejemplo7"))
print(cumple_regla_didactica("ejemplo"))
````

### Proyecto A: resumen sin modificar la entrada

Se supone una lista de montos no negativos. La media entre positivos no está definida si no existen montos positivos.

````python
def resumen_siniestros(montos):
    if any(monto < 0 for monto in montos):
        raise ValueError("Se requieren montos no negativos")
    positivos = [monto for monto in montos if monto > 0]
    return {
        "cantidad": len(montos),
        "total": sum(montos),
        "media_global": calcular_promedio(montos),
        "cantidad_positivos": len(positivos),
        "media_positivos": calcular_promedio(positivos)
    }

datos = [0, 1200, 500, 0, 3500, 800]
print(resumen_siniestros(datos))
print(resumen_siniestros([]))
print(resumen_siniestros([0, 0]))
assert resumen_siniestros(datos)["total"] == 6000
assert resumen_siniestros(datos)["media_global"] == 1000
assert resumen_siniestros(datos)["media_positivos"] == 1500
````

### Proyecto A: clasificación y frecuencias

Mismos intervalos ficticios que en la sección 1.

````python
def clasificar_monto(monto):
    if monto < 0:
        return "No válido"
    if monto < 1000:
        return "Bajo"
    if monto < 5000:
        return "Medio"
    return "Alto"

frecuencias_clase = {"Bajo": 0, "Medio": 0, "Alto": 0}
for monto in [0, 1200, 500, 0, 3500, 800]:
    frecuencias_clase[clasificar_monto(monto)] += 1
print(frecuencias_clase)
assert sum(frecuencias_clase.values()) == 6
assert clasificar_monto(1000) == "Medio"
assert clasificar_monto(5000) == "Alto"
````

### Proyecto B: saldos y número de operaciones

La función reutiliza el formato de archivo definido antes y no escribe sobre la entrada.

````python
def resumir_movimientos(ruta):
    saldo = calcular_saldo_archivo(ruta)
    cantidades = {"Deposito": 0, "Retiro": 0}
    with open(ruta, "r", encoding="utf-8") as archivo:
        archivo.readline()
        for linea in archivo:
            if linea.strip():
                operacion = linea.split("|")[0].strip()
                cantidades[operacion] += 1
    return saldo, cantidades

saldo, cantidades = resumir_movimientos("estado_cuenta.txt")
print("Saldo: ${:.2f}".format(saldo))
print(cantidades)
assert cantidades == {"Deposito": 2, "Retiro": 2}
````

### Proyecto C: sumas de dos dados

La probabilidad teórica de suma 7 se obtiene de 6 pares favorables entre 36 pares equiprobables.

````python
def simular_dados(n, semilla=2026):
    if n <= 0:
        raise ValueError("n debe ser positivo")
    generador = random.Random(semilla)
    frecuencias = {suma: 0 for suma in range(2, 13)}
    for _ in range(n):
        suma = generador.randint(1, 6) + generador.randint(1, 6)
        frecuencias[suma] += 1
    return frecuencias

frecuencias_dados = simular_dados(6000)
print(frecuencias_dados)
print("P estimada de suma 7: {:.4f}".format(frecuencias_dados[7] / 6000))
print("P teórica: {:.4f}".format(6 / 36))
assert sum(frecuencias_dados.values()) == 6000
assert set(frecuencias_dados) == set(range(2, 13))
````

# 1. Python y sus entornos de ejecución

Objetivo: distinguir lenguaje, intérprete, editor, terminal y notebook.

Python es un lenguaje de alto nivel y multiparadigma: permite distintos estilos de programación. En este curso comenzamos con instrucciones secuenciales y expresiones; no hace falta dominar clases para utilizar Python.

Un intérprete ejecuta programas Python. CPython, una implementación habitual, primero compila el código a una representación intermedia llamada bytecode y después la ejecuta. Por ello, «interpretado» no significa «nunca compilado» ni «lee únicamente una línea cada vez». Tampoco permite concluir que cualquier programa Python será siempre más lento o consumirá más memoria que cualquier programa de otro lenguaje.

| Entorno | Uso | Qué debes distinguir |
|---|---|---|
| REPL | Probar instrucciones interactivamente | El indicador `>>>` no forma parte del código que copias |
| Archivo `.py` | Guardar y ejecutar un programa | Una expresión aislada no se imprime automáticamente |
| Notebook `.ipynb` | Combinar explicación, código y resultados | Las variables permanecen en el núcleo entre celdas |
| Editor o IDE | Escribir, revisar y ejecutar programas | El editor no reemplaza al intérprete |
| Terminal | Ejecutar comandos del sistema | No es lo mismo que el REPL |

Para trabajar localmente se necesita Python 3 y un editor; Jupyter permite utilizar notebooks. Un servicio de notebooks en línea puede proporcionar el intérprete sin instalarlo localmente. No dependemos de una imagen externa de instalación.

En una terminal, según tu instalación, puedes comprobar la versión con `python --version`, `python3 --version` o `py --version` y ejecutar un archivo con `python programa.py` (o el comando equivalente disponible). Estos son comandos de terminal, no instrucciones para una celda Python. Para salir del REPL puedes usar `exit()`; también Ctrl+D en muchos entornos Unix o Ctrl+Z seguido de Enter en Windows.

El REPL y Jupyter pueden mostrar el valor de una expresión. En un script usa `print()` para mostrarlo explícitamente. Un notebook con resultados visibles no garantiza que el código actual produzca esos resultados: ejecuta de nuevo tras editarlo.

# 2. Variables, nombres y asignación

Objetivo: asignar valores y elegir nombres válidos y descriptivos.

Una variable es un nombre que hace referencia a un objeto. En `edad = 18`, Python evalúa el lado derecho y asocia el resultado con `edad`. No es necesario declarar previamente un tipo fijo. Los objetos tienen tipo; un mismo nombre puede referirse después a otro tipo de objeto.

Reglas sintácticas básicas para los nombres usados aquí:
- Pueden comenzar con una letra o `_`, pero no con un dígito.
- Pueden continuar con letras, dígitos o `_`; no llevan espacios ni guiones `-`.
- Distinguen mayúsculas y minúsculas: `Edad` y `edad` son nombres distintos.
- No pueden ser palabras reservadas, como `if`, `for` o `def`.

Python admite también ciertos caracteres Unicode en identificadores. Para el curso se recomienda usar nombres descriptivos en minúsculas y separar palabras con guion bajo: `prima_anual`. Esta es una convención de estilo, no una restricción del lenguaje. `Variable` y `_variable` son válidos.

No uses `sum`, `str`, `int`, `input` o `print` como nombres de tus datos: son nombres incorporados y reasignarlos ocultaría su significado habitual. No son todos palabras reservadas. La lista de palabras reservadas depende de la versión; se puede consultar con `keyword.kwlist`.

Fuente: [F2].

````python
prima_anual = 1500
edad = 18
Edad = 20
print(prima_anual, edad, Edad)

x = 5
y = x
x = 8
print("x:", x, "y:", y)
````

En el ejemplo, reasignar `x` no modifica `y`: los enteros son inmutables. No generalices este comportamiento a colecciones mutables compartidas, que se estudiarán después.

Una sentencia expresa una instrucción completa, como `total = 2 + 3`. La parte `2 + 3` es una expresión: se evalúa y produce un valor. El operador `+` actúa sobre sus operandos. El signo `=` pertenece a la sintaxis de asignación; no es una comparación.

El punto y coma permite separar sentencias simples en una misma línea, pero para aprender y depurar se recomienda una instrucción por línea.

````python
base = 5; altura = 8; area = base * altura / 2
print("Área del triángulo:", area)
````

# 3. Tipos de datos y conversiones

Objetivo: reconocer tipos básicos y convertir valores de manera consciente.

| Tipo | Ejemplo | Significado |
|---|---|---|
| `int` | `2026`, `-3` | Entero |
| `float` | `2026.0`, `1.25`, `1e3` | Número de punto flotante; precisión finita |
| `complex` | `2 + 3j` | Número complejo |
| `bool` | `True`, `False` | Valor lógico |
| `str` | `"2026"`, `"Actuaría"` | Cadena de texto |
| `NoneType` | `None` | Ausencia de valor, según el contexto |

`type(valor)` consulta el tipo. Un texto con dígitos no se vuelve número automáticamente: `"17"` es `str`. «Tipado dinámico» no significa ausencia de tipos. Python no convierte automáticamente `"2" + 3` en una suma numérica o en una concatenación, aunque sí admite operaciones numéricas compatibles, como `2 + 3.5`.

Las conversiones habituales son `int()`, `float()`, `complex()` y `str()`. `int(7.9)` trunca hacia cero; no redondea al entero más próximo. `int("3.2")` falla porque el texto no representa un literal entero válido para esa conversión. Para entradas decimales usa punto: `float("3.2")`.

Fuente: [F3].

````python
print(type(2026), type(2026.0), type("2026"), type(True))
print(int(7.9), int(-7.9))
print(float("3.2"), str(25), complex(5))
print("20" + "26")
print(20 + 26)
````

Un complejo tiene partes real e imaginaria. No se convierte directamente con `float(z)` o `int(z)`, incluso si su parte imaginaria es cero. Sí puedes extraer sus partes con `.real` y `.imag`, obtener su módulo con `abs(z)` o convertirlo a texto con `str(z)`. Por eso es incorrecto decir que un complejo «no se puede convertir en otra cosa».

````python
z = 3 + 4j
print("Real:", z.real, "Imaginaria:", z.imag)
print("Módulo:", abs(z))
print("Texto:", str(z))
````

### Complemento: precisión numérica
Los `float` no representan exactamente todos los números reales ni todas las fracciones decimales. Por ejemplo, la suma de `0.1` y `0.2` puede mostrar una pequeña diferencia respecto de `0.3`. No es un error de `print()`.

Mostrar dos decimales no cambia el objeto numérico original. `math.isclose()` permite comparar aproximaciones con tolerancias que deben elegirse según el problema; cerca de cero puede ser necesaria una tolerancia absoluta positiva. Para contabilidad con reglas decimales precisas puede utilizarse `decimal`, tema que no se desarrolla aquí. [F5, F6]

````python
import math
resultado = 0.1 + 0.2
print(resultado)
print(resultado == 0.3)
print(math.isclose(resultado, 0.3))
print(f"Presentación con dos decimales: {resultado:.2f}")
````

# 4. Salida de datos y comentarios

Objetivo: controlar la presentación con print() y documentar el código.

`print()` muestra valores y, de manera predeterminada, separa sus argumentos con un espacio y termina con un salto de línea. Su esquema básico es `print(valor1, valor2, sep=" ", end="\n")`.

| Elemento | Función |

# 5. Operadores aritméticos y precedencia

Objetivo: distinguir división, cociente, residuo y potencia y respetar la precedencia.

| Operador | Ejemplo | Resultado |
|---|---|---|
| `+` | `2 + 5` | `7` |
| `-` | `2 - 5` | `-3` |
| `*` | `5 * 3` | `15` |
| `/` | `15 / 4` | `3.75` |
| `//` | `17 // 3` | `5` |
| `%` | `17 % 3` | `2` |
| `**` | `4 ** 3` | `64` |

`/` es división: incluso `50 / 2` produce `25.0` con estos enteros. `//` es división piso, no residuo; `%` es residuo, no división entera. Para enteros con divisor no nulo se cumple `a == b * (a // b) + (a % b)`.

Con enteros positivos, `//` cuenta grupos completos y `%` los sobrantes. Con negativos, `//` redondea hacia menos infinito: `-7 // 3` es `-3`, no `-2`. Un residuo no nulo tiene el signo del divisor. Con operandos flotantes puede haber aproximaciones y el resultado de `//` puede ser `float`: `7.0 // 2` da `3.0`.

No dividas entre cero: `/`, `//` y `%` con divisor cero generan `ZeroDivisionError`. [F3, F7]

````python
a = 25
b = 6
cociente = a // b
residuo = a % b
print(f"{a} = {b} * {cociente} + {residuo}")
print(-7 // 3, -7 % 3)
print(7.0 // 2, type(7.0 // 2))
````

### Precedencia y asociatividad
Entre las operaciones de este capítulo: primero las agrupaciones con paréntesis; después las potencias; después `*`, `/`, `//`, `%`; finalmente `+` y `-` binarios. Las operaciones del mismo nivel se agrupan de izquierda a derecha, excepto la potencia, que se agrupa de derecha a izquierda.

La regla mnemotécnica PEMDAS no significa «multiplicar siempre antes de dividir»: `12 / 3 * 2` es `(12 / 3) * 2`. Usa paréntesis cuando hagan más clara tu intención.

Cuidado con el signo unario: `-2 ** 2` equivale a `-(2 ** 2)`, mientras que `(-2) ** 2` eleva la base negativa. `2 ** -2` es válido. El símbolo `^` no es potencia en Python.

````python
print(2 + 3 * 4, (2 + 3) * 4)
print(12 / 3 * 2)
print(2 ** 3 ** 2, (2 ** 3) ** 2)
print(-2 ** 2, (-2) ** 2, 2 ** -2)
````

# 6. Asignación compuesta, comparaciones y lógica

Objetivo: seguir actualizaciones y construir expresiones booleanas claras.

Complemento de enlace con las notas de clase previas: asignación compuesta, comparación y lógica.

Para los números usados aquí, `x += y` tiene el mismo efecto de actualización que `x = x + y`. Análogamente existen `-=`, `*=`, `/=`, `//=`, `%=` y `**=`. El tipo puede cambiar: al aplicar `/=` a estos enteros, el resultado pasa a ser flotante.

| Comparación | Significado |
|---|---|
| `==`, `!=` | Igual, diferente |
| `<`, `<=` | Menor, menor o igual |
| `>`, `>=` | Mayor, mayor o igual |

`=` asigna y `==` compara. `2026 == 2026.0` es `True`; `"2026" == 2026` es `False`. Comparar por orden una cadena con un entero, como `"20" < 30`, no es válido en Python 3. [F3]

````python
x = 20
x += 4
print(x)
x /= 3
print(x, type(x))
print(2026 == 2026.0)
print("2026" == 2026)
print("Dato" == "dato")
````

Para operandos booleanos:

| P | Q | P and Q | P or Q |
|---|---|---|---|
| True | True | True | True |
| True | False | False | True |
| False | True | False | True |
| False | False | False | False |

`not True` es `False`; `not False` es `True`. La precedencia lógica es `not`, luego `and`, luego `or`; las comparaciones tienen mayor precedencia que estos operadores lógicos. Puedes usar paréntesis para mejorar la legibilidad.

Complemento: `and` y `or` trabajan con cortocircuito; no siempre evalúan el segundo operando. Con operandos no booleanos pueden devolver uno de sus operandos, no necesariamente un `bool`. En este curso comenzamos con comparaciones que producen valores booleanos. [F3]

````python
edad = 20
prima = 1500
cumple = (edad >= 18) and (edad <= 65) and (prima > 0)
print(cumple)

divisor = 0
print((divisor != 0) and (10 / divisor > 2))  # La división no se evalúa.
````

# 8. Indentación y decisiones básicas

Objetivo: identificar bloques y leer decisiones sencillas sin confundir errores sintácticos y lógicos.

La indentación agrupa instrucciones en bloques. En una decisión, los dos puntos `:` abren el bloque; las instrucciones del mismo bloque deben tener sangría coherente. Usa cuatro espacios por nivel como convención de estilo y evita mezclar espacios y tabuladores.

`if` ejecuta su bloque si la condición es verdadera. Un `elif` prueba otra condición si las anteriores fueron falsas. `else` cubre el caso restante. Las palabras `if`, `elif` y `else` del mismo conjunto se alinean. No existe un `endif` en Python.

El ejemplo `if/elif/else` que aparecía en las notas era sintácticamente válido, aunque el texto lo anunciaba como un error. Aquí se presenta como ejemplo correcto; los errores intencionales se muestran en el capítulo 14. [F2, F7]

````python
x = -1
if x > 0:
    print("x es positivo")
elif x < 0:
    print("x es negativo")
else:
    print("x es cero")
print("Esta instrucción está fuera de la decisión.")
````

````python
x = 2
if x > 0:
    print("Valor positivo original:", x)
    x = x - 1
print("Valor actual:", x)

x = -2
if x > 0:
    x = x - 1
print("Valor actual con entrada negativa:", x)
````

La indentación incorrecta puede generar un error de sintaxis o cambiar la lógica sin producir una excepción. Por ejemplo, mover el último `print()` al interior del `if` cambia cuándo se muestra el resultado. Lee los bloques, no solo las líneas aisladas.

# 9. Funciones y módulos matemáticos

Objetivo: usar funciones incorporadas y math respetando dominios matemáticos.

Una función se llama con paréntesis: `abs(-5)`. Un módulo organiza funciones, clases y otros nombres; muchos módulos están escritos en archivos `.py`, pero también hay módulos incorporados o compilados, de modo que no todo módulo es necesariamente un archivo Python visible.

`abs()` es incorporada y no requiere importación. Para usar `floor`, `ceil` o `log` del módulo `math`, escribe primero `import math` y después `math.nombre(...)`. Esto no es lo mismo que definir una función propia. [F5]

### Piso, techo y truncamiento
Para números reales finitos:
\[
\lfloor x\rfloor=\max\{k\in\mathbb{Z}:k\leq x\},\qquad
\lceil x\rceil=\min\{k\in\mathbb{Z}:x\leq k\}.
\]
Ambas funciones toman valores en los enteros \(\mathbb{Z}\), no solo en los naturales. `floor` va hacia menos infinito, `ceil` hacia más infinito e `int(float)` trunca hacia cero. No son sinónimos de redondeo al entero más cercano.

````python
import math
print("Valor absoluto:", abs(-34))
print("Piso y techo de 6.87:", math.floor(6.87), math.ceil(6.87))
print("Piso y techo de -1.5:", math.floor(-1.5), math.ceil(-1.5))
print("Truncamiento de -1.5:", int(-1.5))
````

### Logaritmos y dominio
`math.log(x)` calcula \(\ln(x)\) para \(x>0\). `math.log(x, b)` calcula el logaritmo en base \(b\) para \(b>0\), \(b\neq1\). No todas las entradas numéricas son matemáticamente válidas. `math.log(0)` y `math.log(-1)` generan `ValueError`.

Complemento: `math.sqrt(x)` requiere \(x\geq0\) en el dominio real; `math.exp(x)` calcula \(e^x\). Para funciones complejas existe `cmath`, fuera del desarrollo central de este cuaderno. [F5]

````python
import math
print(math.log(1))
print(math.log(math.e))
print(math.log(8, 2))
print(math.sqrt(25))
````

# 10. Cadenas: creación e inmutabilidad

Objetivo: distinguir literales, concatenación y creación de nuevas cadenas.

Una cadena `str` es una secuencia inmutable de puntos de código Unicode. Para los ejemplos sencillos de este curso, hablamos de caracteres; en textos con combinaciones de marcas o algunos emojis, un carácter visual puede ocupar más de una posición.

Puedes usar comillas simples, dobles o triples. Las triples permiten conservar saltos de línea reales. Una barra invertida al final de una línea del código permite continuar la instrucción, pero no agrega por sí sola un salto al texto. Por eso no es equivalente a una cadena multilínea.

Las cadenas crudas, con prefijo `r`, permiten conservar barras invertidas sin interpretar las secuencias de escape usuales. No pueden terminar en un número impar de barras invertidas. [F2, F3]

````python
mensaje = "Aprender Python requiere práctica."
multilinea = """Primera línea
Segunda línea"""
una_linea = ("Primera parte "
            "segunda parte")
print(multilinea)
print(una_linea)
print(r"C:\notas\tema1")
````

### Concatenación y repetición
`+` concatena cadenas; `*` con un entero repite una cadena. Los literales adyacentes se concatenan automáticamente: `"Py" "thon"`; esto no se aplica a una variable seguida de un literal.

`" ".join(partes)` une cadenas con un separador. Todos los elementos deben ser cadenas: no convierte números automáticamente. La lista de este ejemplo se usa como contenedor de textos; su estudio detallado se deja para después.

````python
nombre = "Ana"
apellido = "López"
print(nombre + " " + apellido)
print(3 * "UMAR" + " Huatulco")
print("Py" "thon")
print(" ".join([nombre, apellido]))
````

### Inmutabilidad no significa que el nombre quede congelado
No puedes asignar un carácter con `texto[0] = "J"`. Sí puedes construir otra cadena y asociarla al mismo nombre. Los métodos de transformación, como `replace()`, devuelven un resultado sin modificar la cadena original. No es necesario describir cada resultado como una copia física obligatoria: lo importante es que el objeto original no se modifica.

````python
texto = "Python"
nuevo_texto = "J" + texto[1:]
print("Original:", texto)
print("Nuevo:", nuevo_texto)
texto = texto.upper()
print("Nombre reasignado:", texto)
````

# 11. Índices y rebanadas

Objetivo: localizar caracteres y extraer subcadenas sin confundir extremos.

En `palabra = "Python"`, los índices son:

| Carácter | P | y | t | h | o | n |
|---|---|---|---|---|---|---|
| Índice desde el inicio | 0 | 1 | 2 | 3 | 4 | 5 |
| Índice desde el final | -6 | -5 | -4 | -3 | -2 | -1 |

`palabra[2]` es `"t"`, no el carácter de posición cero. `-0` es igual a `0`. Un índice individual fuera del intervalo válido genera `IndexError`; la indexación de una cadena vacía también falla.

Una rebanada tiene forma `s[inicio:fin:paso]`: incluye el inicio y excluye el fin. Sin paso explícito, avanza de uno en uno. Con paso positivo, omitir inicio o fin lleva al principio o al final. Los valores predeterminados cambian para un paso negativo. Los límites fuera de rango se ajustan; una rebanada puede devolver `""` sin error. El paso no puede ser cero.

Fuente: [F3, F7].

````python
palabra = "Python"
print(palabra[0], palabra[2], palabra[-1], palabra[-6])
print(palabra[0:2], palabra[2:5], palabra[-2:])
print(palabra[:2] + palabra[2:])
print(repr(palabra[42:]), palabra[4:42])
print("Longitud:", len(palabra))
print("Complemento, salto de dos:", palabra[::2])
print("Complemento, inversión:", palabra[::-1])
````

Para una cadena no vacía `s`, el último índice no negativo es `len(s) - 1`, no `len(s)`. La función `len()` devuelve un entero y cuenta las posiciones de la secuencia. Las rebanadas no modifican la cadena original.

# 12. Formato de cadenas y presentación numérica

Objetivo: presentar valores con format() y f-strings sin alterar el cálculo.

El método `str.format()` utiliza campos de reemplazo delimitados por llaves `{}`, no por corchetes `[]`. Los campos pueden ser automáticos, posicionales desde cero o nombrados. No mezcles campos automáticos con índices numéricos explícitos en la misma cadena.

Una f-string comienza con `f` antes de la comilla y evalúa expresiones dentro de las llaves en el momento en que se construye la cadena. Ambos mecanismos funcionan en tiempo de ejecución. `format()` no evalúa expresiones Python arbitrarias dentro del campo, pero sus argumentos sí pueden ser expresiones: `"{}".format(2 + 3)`. No es obligatorio calcularlas previamente en otra variable. [F2, F4, F8]

````python
x = 2
y = 3
print("La suma de {} + {} = {}".format(x, y, x + y))
print("{1}, {0}".format("Ana", "López"))
print("Nombre: {nombre}; edad: {edad}".format(nombre="Ana", edad=20))
print(f"La suma de {x} + {y} = {x + y}")
print(f"Llaves literales: {{ejemplo}}")
````

### Precisión, porcentaje y alineación
En `{:.2f}` o `{valor:.2f}`, `.2` indica dos cifras después del punto y `f` notación fija. El resultado es una cadena; no cambia la precisión del número almacenado. El porcentaje multiplica por cien al presentar: `0.075` con `.1%` aparece como `7.5%`.

| Formato | Uso |
|---|---|
| `d` | Entero decimal |
| `b`, `o`, `x`, `X` | Enteros en binario, octal, hexadecimal |
| `.2f` | Dos posiciones decimales |
| `.2e` | Notación científica con dos decimales en la mantisa |
| `.4g` | Precisión general de cuatro cifras significativas |
| `.1%` | Porcentaje con un decimal |
| `<12`, `>12`, `^12` | Alinear a izquierda, derecha o centro, con ancho mínimo 12 |

El ancho es mínimo: no corta textos largos. El formato `g` suele usar seis cifras significativas si no se especifica otra precisión; no garantiza conservar todos los dígitos originales. Estos formatos presentan valores, no cambian la base del objeto entero ni convierten una tasa en otra tasa. [F8]

````python
precio = 15.5
cantidad = 2
total = precio * cantidad
print("Precio: ${:.2f}; cantidad: {}; total: ${:.2f}".format(precio, cantidad, total))
print(f"Tasa: {0.075:.1%}")
print(f"Científica: {1234.5678:.2e}")
print(f"General: {12345.6:g}")
print("Binario: {:b}; octal: {:o}; hexadecimal: {:X}".format(10, 10, 255))
print(f"|{'Alumno':^20}|{'Calificación':>14}|")
print(f"|{'Jorge':<20}|{9.564:>14.2f}|")
````

# 13. Métodos de cadenas

Objetivo: elegir métodos de texto según la tarea y reconocer el tipo devuelto.

Un método se llama mediante un objeto: `texto.lower()`. Una función incorporada como `len(texto)` recibe el objeto como argumento. No son exactamente la misma forma de llamada.

Los métodos de transformación de `str` no modifican la cadena original. Para conservar el resultado utiliza otro nombre o reasigna el original: `texto = texto.strip()`.

### Limpieza y mayúsculas
| Método | Comportamiento |
|---|---|
| `lower()` | Conversión a minúsculas |
| `upper()` | Conversión a mayúsculas |
| `capitalize()` | Primer carácter en mayúscula y resto en minúsculas |
| `title()` | Aplica reglas de título basadas en grupos de letras; no es un corrector de nombres propios |
| `casefold()` | Conversión orientada a comparación sin distinción de mayúsculas |
| `strip()` | Quita blancos en ambos extremos, no los interiores |
| `lstrip()`, `rstrip()` | Quita blancos del extremo izquierdo o derecho |
| `replace(viejo, nuevo)` | Reemplaza coincidencias; no requiere modificar el original |

Con un argumento, `strip(caracteres)` interpreta un conjunto de caracteres a quitar de los extremos, no una subcadena exacta. [F3]

````python
nombre = "  ana lópez  "
limpio = nombre.strip()
print(repr(nombre), repr(limpio))
print(limpio.upper(), limpio.title(), limpio.capitalize())
print("Straße".casefold())
print("UMAR Huatulco".replace("Huatulco", "Campus Huatulco"))
print(repr("  dos  palabras  ".strip()))
````

### Búsqueda, separación y unión
| Método | Resultado |
|---|---|
| `find(subcadena)` | Índice de la primera coincidencia o `-1` si no existe |
| `count(subcadena)` | Número de coincidencias no superpuestas |
| `startswith(prefijo)` | Booleano |
| `endswith(sufijo)` | Booleano |
| `split(separador)` | Lista de fragmentos |
| `separador.join(fragmentos)` | Cadena formada a partir de elementos que deben ser cadenas |

`find()` devuelve `0` si encuentra la coincidencia al principio: no confundas cero con ausencia. `split()` sin argumento agrupa blancos consecutivos y omite fragmentos vacíos de los extremos; `split(",")` conserva fragmentos vacíos entre separadores consecutivos. [F3]

````python
texto = "hola mundo"
print(texto.find("hola"), texto.find("mundo"), texto.find("Python"))
print("banana".count("ana"))
print("reporte.csv".endswith(".csv"))
print("ACT-001".startswith("ACT-"))
print("  Ana   López  ".split())
print("uno,,tres".split(","))
print(" | ".join(["Actuaría", "UMAR", "Huatulco"]))
````

### Clasificación y presentación
`isdigit()`, `isalpha()` e `isspace()` requieren al menos un carácter para devolver `True`; una cadena vacía devuelve `False` en los tres casos. Trabajan con categorías Unicode: `isdigit()` no garantiza que `int()` acepte la cadena. Además, `"-3"` y `"3.2"` no son cadenas formadas solo por dígitos, aunque representen números válidos para ciertas conversiones.

| Método | Uso |
|---|---|
| `isdigit()` | ¿Todos los caracteres son dígitos? |
| `isalpha()` | ¿Todos son letras? Los espacios hacen que sea falso |
| `isspace()` | ¿Todos son caracteres de espacio en blanco? |
| `zfill(ancho)` | Rellena con ceros a la izquierda; respeta un signo inicial |
| `center(ancho, relleno)` | Centra con un carácter de relleno |
| `expandtabs(tabsize)` | Expande tabulaciones hasta posiciones de tabulación |
| `encode(codificacion)` | Devuelve `bytes`, no `str` |

`"hola".center(10, "*")` produce `"***hola***"` (longitud 10). `expandtabs()` no sustituye cada tabulador por una cantidad constante de espacios: depende de la posición actual y del tamaño de tabulación, 8 por defecto. `encode()` es una ampliación sobre representación de texto, no un requisito para imprimir acentos. [F3]

````python
print("123".isdigit(), "-3".isdigit(), "3.2".isdigit(), "".isdigit())
print("Ana".isalpha(), "Ana López".isalpha())
print(" 	".isspace())
print("42".zfill(5), "-42".zfill(5))
print("hola".center(10, "*"))
print(repr("hola	mundo".expandtabs()))
print("Actuaría".encode("utf-8"))
````

### 15.2. Resumen ficticio de primas
Supuestos: tres pólizas con la misma prima y un recargo ilustrativo del 5% aplicado una sola vez al subtotal. La tasa es `0.05`, no `5`. No se trata de un cálculo actuarial de tarifa. La presentación monetaria tiene dos decimales; no se establece una política profesional de redondeo.

````python
cantidad = 3
prima_unitaria = 1250.0
tasa_recargo = 0.05
subtotal = cantidad * prima_unitaria
recargo = subtotal * tasa_recargo
total = subtotal + recargo
print(f"Subtotal: ${subtotal:.2f}")
print(f"Recargo ({tasa_recargo:.1%}): ${recargo:.2f}")
print(f"Total: ${total:.2f}")
````

### 15.3. Paquetes y capacidad
Para 87 expedientes con capacidad de 10 por paquete, `//` cuenta paquetes completos, `%` cuenta expedientes sobrantes y el techo del cociente cuenta paquetes necesarios para guardar todos. Los datos del ejemplo son positivos; la capacidad no puede ser cero.

````python
import math
expedientes = 87
capacidad = 10
completos = expedientes // capacidad
sobrantes = expedientes % capacidad
necesarios = math.ceil(expedientes / capacidad)
print(f"Completos: {completos}; sobrantes: {sobrantes}; necesarios: {necesarios}")
````

### 15.4. Crecimiento compuesto (complemento)
Con capital inicial \(C\), tasa efectiva por periodo \(i\) y \(n\) periodos, un modelo elemental sin aportaciones ni retiros es \(M=C(1+i)^n\). Debes usar periodos compatibles: una tasa anual con años, por ejemplo. No agregamos inflación, comisiones o impuestos. Este modelo sirve para practicar expresiones y formato, no como recomendación de inversión.

````python
capital = 10000.0
tasa = 0.05
periodos = 2
monto = capital * (1 + tasa) ** periodos
print(f"Monto al final de {periodos} periodos: ${monto:.2f}")
````

# 18. Panorama de colecciones y objetos

Si se vuelve a asignar una clave, su valor se actualiza. Una tupla no permite reemplazar sus elementos, aunque puede contener un objeto mutable cuyo contenido cambie. El conjunto vacío se escribe `set()`; `{}` crea un diccionario vacío.

Los operadores `in` y `not in` consultan pertenencia; en cadenas buscan subcadenas y en diccionarios consultan claves. `is` e `is not` comparan identidad de objetos, no igualdad general de valores. Para comparar los números o textos de estos apuntes usa `==`, no `is`.

En programación orientada a objetos, una clase describe un tipo de objeto; un objeto puede exponer datos mediante atributos y comportamiento mediante métodos. Usar `texto.upper()` permite observar un método sin que aún sea necesario definir una clase. Estos conceptos se ampliarán cuando corresponda en el programa del curso. [F3]
